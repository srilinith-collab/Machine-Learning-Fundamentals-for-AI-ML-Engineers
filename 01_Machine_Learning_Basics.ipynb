## 1. What is Machine Learning?

**Machine Learning (ML)** is a branch of Artificial Intelligence in which a system learns patterns from data and improves its performance on a task **without being explicitly programmed** with fixed rules.



## 2. Why Machine Learning?

- Some problems are too complex for hand-written rules (e.g., recognizing handwriting, predicting customer churn).
- Data is growing rapidly, and ML can find patterns humans might miss.
- ML systems can **adapt and improve** as more data becomes available.
- Automates repetitive decision-making tasks (fraud detection, recommendations, forecasting).

## 3. AI vs ML vs Deep Learning

| Concept | Description |
|---|---|
| **Artificial Intelligence (AI)** | The broad science of building systems that mimic human intelligence (reasoning, perception, decision-making). |
| **Machine Learning (ML)** | A subset of AI where systems learn patterns from data instead of following explicit rules. |
| **Deep Learning (DL)** | A subset of ML that uses multi-layered neural networks to learn complex patterns, especially from images, text, and audio. |

**Relationship:** `AI ⊃ ML ⊃ Deep Learning`

## 4. Traditional Programming vs Machine Learning

| Traditional Programming | Machine Learning |
|---|---|
| `Rules + Data → Output` | `Data + Output → Rules (Model)` |
| Human writes the logic | Algorithm learns the logic from data |
| Works well for well-defined problems | Works well for complex, pattern-based problems |

Example: To detect spam emails traditionally, you'd write hundreds of if-else rules. In ML, you show the algorithm thousands of labeled spam/non-spam emails, and it learns the pattern itself.

## 6. Supervised Learning

In **Supervised Learning**, the model learns from **labeled data** — each input example (features) has a known correct output (target/label).

The model learns a mapping: `Features → Target`

Common tasks: **Classification** and **Regression** (both covered below).

Example here: Using `Quantity` and `UnitPrice` (features) to predict `TotalPrice` (target) — a regression problem.

## 7. Unsupervised Learning

In **Unsupervised Learning**, the data has **no labels**. The model tries to find hidden structure or patterns in the data on its own.

Common task: **Clustering** — grouping similar data points together (e.g., grouping customers by purchasing behavior).

## 8. Semi-Supervised Learning

**Semi-Supervised Learning** uses a **small amount of labeled data** combined with a **large amount of unlabeled data** during training.

- Useful when labeling data is expensive or time-consuming (e.g., manually labeling millions of customer transactions as "fraud" vs "not fraud").
- The model first learns structure from the unlabeled data, then uses the few labeled examples to refine its understanding.

## 9. Reinforcement Learning

**Reinforcement Learning (RL)** is different from the above — there's no fixed dataset of features/labels. Instead:

- An **agent** interacts with an **environment**.
- It takes **actions**, receives **rewards or penalties**, and learns a **policy** to maximize cumulative reward over time.

Example: An RL agent could learn an optimal **dynamic pricing strategy** for our retail products — it tries different prices (actions), observes sales/profit (reward), and learns which pricing strategy maximizes revenue.

## 12. Clustering

**Clustering** is an unsupervised learning task that groups similar data points together **without predefined labels**.

Example: Group customers based on their **total spend** and **purchase frequency** (a simple customer segmentation).

## 13. Core Machine Learning Terminology

Let's tie everything above back to the core vocabulary of ML, using our regression example (predicting `TotalPrice`) as reference.

### 13.1 Features

**Features** are the input variables used to make predictions.

Example: `Quantity`, `UnitPrice`

### 13.2 Target

The **Target** (also called the label) is the output variable the model is trying to predict.

Example: `TotalPrice` (regression) or `ValueClass` (classification)

### 13.3 Training

**Training** is the process of showing the model data so it can learn patterns.

Example: `model.fit(X_train, y_train)`

### 13.4 Prediction

**Prediction** is using the trained model to estimate the output for new/unseen data.

Example: `model.predict(X_test)`

### 13.5 Model

The **Model** is the mathematical representation learned from data that maps features to the target.

Example: `LinearRegression()`, `LogisticRegression()`, `KMeans()` objects

### 13.6 Parameters

**Parameters** are internal values the model **learns automatically** during training.

Example: Regression coefficients (`reg_model.coef_`), intercept (`reg_model.intercept_`)

### 13.7 Hyperparameters

**Hyperparameters** are configuration values **set before training** by the user, not learned from data.

Example: `n_clusters=3` in KMeans, `test_size=0.2` in train/test split
